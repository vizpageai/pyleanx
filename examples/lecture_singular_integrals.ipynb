# Singular Integrals I: Covering Lemmas and the Maximal Function
### Stein, *Singular Integrals*, ch. I §1 — proved step by step and checked by Lean

The Hardy–Littlewood maximal function
$$Mf(x) = \sup_{r>0} \frac{1}{|I(x,r)|}\int_{I(x,r)} |f(y)|\,dy, \qquad I(x,r) = [x-r, x+r],$$
is the entry point of real-variable harmonic analysis (Stein, *Singular Integrals and Differentiability
Properties of Functions*, ch. I; Tao's notes on the maximal inequality). Its basic estimate is the
**weak type (1,1) inequality** $\alpha\,|\{Mf>\alpha\}| \le C\,\|f\|_1$.

Mathlib (the library Lean uses) does **not** contain this theorem. In this notebook it is proved from
scratch, as a sequence of **PyLeanX written proofs**: every step is a sentence with its reason — a
general Mathlib fact (Vitali's selection, compactness, inner regularity, subadditivity of measure, …) or
an earlier theorem of the chapter — and Lean checks every step. Nothing is assumed.

| # | Theorem | Proof uses |
|---|---|---|
| 1 | Vitali covering lemma for finitely many intervals | Mathlib's Vitali selection, finiteness |
| 2 | $\sum_j \int_{I_j}|f| \le \|f\|_1$ for disjoint intervals | integral over a disjoint union |
| 3 | $\alpha\,|\bigcup I_i| \le 4\|f\|_1$ if the average of $|f|$ over each $I_i$ exceeds $\alpha$ | 1, 2, subadditivity |
| 4 | $\alpha\,|K| \le 4\|f\|_1$ for compact $K \subseteq \{Mf>\alpha\}$ | choice, compactness, 3 |
| 5 | $\alpha\,|E| \le 4\|f\|_1$ for measurable $E \subseteq \{Mf>\alpha\}$ of finite measure | inner regularity, 4 |

($\{Mf>\alpha\}$ is written out as its definition: the points $x$ having an interval $[x-\rho,x+\rho]$ on
which the average of $|f|$ exceeds $\alpha$. The constant 4 comes from Vitali with dilation factor 4.)

## 1. The idea in pictures

**Vitali's covering lemma.** From a finite family of intervals, choose greedily the longest interval,
discard those meeting it, and repeat. The chosen intervals are disjoint, and every interval of the family
lies in the 3-fold dilate of a chosen one (Mathlib's version allows any factor $>3$; we use 4).

## 2. The proofs, as written

Each theorem is printed as Lean checks it. Run the next cell to see the five proofs; the one after it

**In the book (Stein, Theorem I.1(b)).** *Let $E_\alpha = \{Mf > \alpha\}$ and let $K \subseteq E_\alpha$ be
compact. Each $x\in K$ is the centre of an interval $B_x$ with $\int_{B_x}|f| > \alpha|B_x|$. By compactness
finitely many $B_x$ cover $K$; by Vitali there is a disjoint subfamily $B_1,\dots,B_m$ with
$|K| \le 4\sum|B_j| \le \frac4\alpha\sum\int_{B_j}|f| \le \frac4\alpha\|f\|_1$. By inner regularity the bound
passes to $E_\alpha$.* ∎ — Theorems 1–5 are exactly these sentences.

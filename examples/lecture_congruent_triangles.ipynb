## 1. What does *congruent* mean?

Two triangles are **congruent**, written $\triangle ABC \cong \triangle DEF$, when one can be moved
onto the other by a rigid motion (slide, turn, flip). Then the **corresponding** sides and angles are equal:

$$AB = DE,\quad BC = EF,\quad CA = FD,\qquad \angle A = \angle D,\quad \angle B = \angle E,\quad \angle C = \angle F.$$

**The order of the letters matters:** $\triangle ABC \cong \triangle DEF$ pairs $A\leftrightarrow D$, $B\leftrightarrow E$, $C\leftrightarrow F$.

We do not have to check all six equalities. Three well-chosen ones are enough:

| Criterion | What you need | PyLeanX reason |
|---|---|---|
| **SSS** | three pairs of equal sides | `geo.sss` |
| **SAS** | two sides and the angle *between* them | `geo.sas` |
| **ASA** | two angles and the side *between* them | `geo.asa` |
| **AAS** | two angles and a side *not* between them | `geo.aas` |
| **HL** | right triangles: hypotenuse and one leg | `geo.hl` |

Once two triangles are congruent, **CPCTC** gives any corresponding part:

| To get… | in $\triangle abc \cong \triangle a'b'c'$ | PyLeanX reason |
|---|---|---|
| $ab = a'b'$, $bc = b'c'$, $ca = c'a'$ | a side | `geo.cpctc_ab`, `geo.cpctc_bc`, `geo.cpctc_ca` |
| $\angle cab = \angle c'a'b'$ (angle at the 1st vertex) | an angle | `geo.cpctc_angle_a` |
| $\angle abc = \angle a'b'c'$ (angle at the 2nd vertex) | an angle | `geo.cpctc_angle_b` |
| $\angle bca = \angle b'c'a'$ (angle at the 3rd vertex) | an angle | `geo.cpctc_angle_c` |

### Why not SSA?

Two sides and an angle that is **not** between them do *not* fix the triangle. In the figure, both
triangles have $AB = AB$, $BC = BC'$ and the same angle at $A$, yet they are clearly different.

````python
pts = {"A": (0, 0), "B": (3.2, 2.4), "C": (5.6, 0), "C'": (2.4, 0)}
show(draw(pts, ["AB", "BC", "BC'", "AC"], fills=[("ABC", "fill1"), ("ABC'", "fill2")],
          ticks={"BC": 2, "BC'": 2}, arcs={"BAC": 1}, title="SSA: two different triangles"))
````

## 2. Writing a proof in PyLeanX

A proof is a list of steps. Each step is a **statement** and a **reason**:

```python
pl.claim(<statement>, because=<reason>)   # one line of the two-column proof
pl.conclude(because=<reason>)             # the last line: "what we wanted to prove"
```

| Write this | to say |
|---|---|
| `dist(A, B) == dist(D, E)` | $AB = DE$ |
| `angle(B, A, C) == angle(E, D, F)` | $\angle BAC = \angle EDF$ (the angle **at the middle letter**) |
| `geo.congruent((A, B, C), (D, E, F))` | $\triangle ABC \cong \triangle DEF$ |
| `pi()` | $180^\circ$ (angles are measured in radians: $\pi = 180^\circ$) |
| `because="hAB"` | a *given* (a hypothesis, by its name) |
| `because=geo.sas` | a theorem from the geometry library |
| `because=["hAB", geo.dist_comm]` | several reasons together |

Reasons you will need today, besides the criteria and CPCTC above:

| Reason | Meaning |
|---|---|
| `geo.dist_comm` | $AB = BA$ |
| `geo.angle_comm` | $\angle ABC = \angle CBA$ |
| `geo.midpoint_equal_halves` | if $M$ is the midpoint of $BC$, then $BM = CM$ |
| `geo.midpoint_straight` | if $M$ is the midpoint of $BC$ (and $B \ne C$), then $\angle BMC = 180^\circ$ |
| `geo.vertical_angles` | vertical angles are equal |
| `geo.linear_pair` | angles on a straight line add up to $180^\circ$ |
| `geo.isosceles_base_angles` | base angles of an isosceles triangle are equal |
| `geo.angle_sum` | the angles of a triangle add up to $180^\circ$ |
| `geo.distinct_ab`, `geo.distinct_bc`, … | the vertices of a triangle are different points |

Lean fills in small routine gaps itself (for example, it knows $AC = AC$ and simple algebra),
but **every geometric fact must be backed by a reason.**

## 3. Worked example 1: the kite (SAS)

**Explore.** Move the sliders. $AB = AD$ and $\angle BAC = \angle DAC$ stay true. What do you notice about $BC$ and $DC$?

````python
show(figure_kite())
````

````python
explore_kite()
````

**Theorem.** If $AB = AD$ and $\angle BAC = \angle DAC$, then $BC = DC$.

| # | Statement | Reason |
|---|---|---|
| 1 | $BA = DA$ | given $AB = AD$, and $AB = BA$ |
| 2 | $\triangle BAC \cong \triangle DAC$ | **SAS**: $BA = DA$, $\angle BAC = \angle DAC$, $AC = AC$ |
| 3 | $CB = CD$ | **CPCTC** (the third sides: from $C$ back to $B$, and to $D$) |
| 4 | $BC = DC$ | $XY = YX$ |

Notice the order in step 2: the angle $\angle BAC$ sits at the **middle** letter $A$, between the sides
$BA$ and $AC$. That is exactly what SAS needs. In PyLeanX:

````python
kite_proof = kite(
    pl.claim(dist(B, A) == dist(D, A), because=geo.dist_comm),
    pl.claim(geo.congruent((B, A, C), (D, A, C)), because=geo.sas,
             note="SAS: BA = DA, ∠BAC = ∠DAC, AC = AC"),
    pl.claim(dist(C, B) == dist(C, D), because=geo.cpctc_ca),
    pl.conclude(because=geo.dist_comm),
)
check(kite_proof)
````

## 4. Worked example 2: the median of an isosceles triangle (SSS)

**Explore.** $AB = AC$ and $M$ is the midpoint of $BC$. Watch $\angle AMB$ and $\angle AMC$ as you move the sliders.

````python
show(figure_isosceles())
````

````python
explore_isosceles()
````

**Theorem.** In $\triangle ABC$ with $AB = AC$, the median $AM$ to the base is perpendicular to the base: $\angle AMB = 90^\circ$.

| # | Statement | Reason |
|---|---|---|
| 1 | $BM = CM$ | $M$ is the midpoint of $BC$ |
| 2 | $\triangle BMA \cong \triangle CMA$ | **SSS**: $BM = CM$, $MA = MA$, $AB = AC$ |
| 3 | $\angle BMA = \angle CMA$ | **CPCTC** |
| 4 | $B \ne C$ | vertices of a triangle are distinct |
| 5 | $\angle BMC = 180^\circ$ | $M$ lies on segment $BC$ |
| 6 | $\angle AMB + \angle AMC = 180^\circ$ | linear pair |
| 7 | $\angle AMB = 90^\circ$ | two equal angles adding up to $180^\circ$ |

````python
median_proof = isosceles_median(
    pl.claim(dist(B, M) == dist(C, M), because=geo.midpoint_equal_halves),
    pl.claim(geo.congruent((B, M, A), (C, M, A)), because=geo.sss,
             note="SSS: BM = CM, MA = MA, AB = AC"),
    pl.claim(angle(B, M, A) == angle(C, M, A), because=geo.cpctc_angle_b),
    pl.claim(B != C, because=geo.distinct_bc("hABC")),
    pl.claim(angle(B, M, C) == pi(), because=geo.midpoint_straight),
    pl.claim(angle(A, M, B) + angle(A, M, C) == pi(), because=geo.linear_pair),
    pl.conclude(because=geo.angle_comm, note="two equal angles adding up to 180°"),
)
check(median_proof)
````

## 5. What happens when a reason is wrong?

A common mistake: in the kite, a student claims the triangles are congruent by **SSS**.
But SSS needs $BC = DC$, which is exactly what we are trying to prove! Lean refuses:

````python
wrong = kite(
    pl.claim(dist(B, A) == dist(D, A), because=geo.dist_comm),
    pl.claim(geo.congruent((B, A, C), (D, A, C)), because=geo.sss),   # ✗ circular
    pl.claim(dist(C, B) == dist(C, D), because=geo.cpctc_ca),
    pl.conclude(because=geo.dist_comm),
)
check(wrong)
````

Lean points at **step 2**: nothing we know so far gives the third pair of sides. A proof
assistant never accepts "it looks right in the picture."

## 7. Summary

* Congruence needs **three** well-chosen equal parts: SSS, SAS, ASA, AAS, HL — never SSA.
* The included angle in SAS (and the included side in ASA) is at the **middle letter**.
* After proving congruence, **CPCTC** gives any other pair of corresponding parts.
* A proof is only a proof when every step has a valid reason; Lean checks each one.

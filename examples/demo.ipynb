# PyLeanX demo
Statements are typeset with LaTeX; **Lean 4 decides** whether proofs are correct.

Each `verify()` runs Lean (about 30–60 s on this machine; a repeat of an unchanged proof is cached).

```python
import pyleanx as pl
x, y = pl.symbols("x y", pl.Real)
θ, ε = pl.symbols("θ ε", pl.Real)
```

## Expressions display as math

```python
(x + y)**2 == x**2 + 2*x*y + y**2
```

```python
pl.Forall([θ], pl.Exists([ε], pl.And(ε > 0, ε < θ**2 + 1)))
```

```python
abs(θ - 1) <= ε
```

## A theorem via the decorator
Evaluating it shows the typeset statement and the generated Lean (click *Lean source*).

```python
@pl.theorem(proof=pl.ring())
def square_identity(x: pl.Real, y: pl.Real):
    return (x + y)**2 == x**2 + 2*x*y + y**2

square_identity
```

```python
square_identity.verify()   # runs Lean + axiom audit
```

## Hypotheses and Greek letters

```python
@pl.theorem(proof=pl.positivity(), assumptions=lambda α, x: {"hα": α > 0})
def scaled_square(α: pl.Real, x: pl.Real):
    return α * x**2 + α > 0

scaled_square
```

```python
scaled_square.verify()
```

## A false claim is rejected

```python
bad = pl.Theorem("false_claim", [x], conclusion=x**2 < 0, proof=pl.positivity())
bad.verify()
```

## LaTeX source for papers

```python
print(pl.latex(scaled_square))
```
